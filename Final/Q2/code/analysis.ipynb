```python
import os
import pandas as pd
import numpy as np
```

Number of classes

```python
dir_path = "text"

file_count = 0
for file in os.listdir(dir_path):
    full_path = os.path.join(dir_path, file)
    if os.path.isfile(full_path):
        file_count += 1

print(f'Number of files: {file_count}')
```

Class dates

```python

for file in os.listdir(dir_path):
    full_path = os.path.join(dir_path, file)
    if os.path.isfile(full_path):
        print(file)
```

Median class attendance

```python
attendance_list = []
total_attendance = 0
for file in os.listdir(dir_path):
    full_path = os.path.join(dir_path, file)
    if os.path.isfile(full_path):
        df = pd.read_csv(full_path)
        
        length = len(df)
        print(file)
        print(length)
        total_attendance +=  length
        attendance_list.append(length)

avg = total_attendance / file_count
median = np.median(attendance_list)

print(f'Average: {avg}')
print(f'Median: {median}')
```

Date with lowest attendance

```python
min_attendance = 100
min_date = ""
for file in os.listdir(dir_path):
    full_path = os.path.join(dir_path, file)
    if os.path.isfile(full_path):
        df = pd.read_csv(full_path)
        attendance = len(df)
        if attendance <= min_attendance:
            min_attendance = attendance
            min_date = file

print(min_date)
print(min_attendance)
```

Date with highest attendance

```python
max_attendance = 0
max_date = ""
for file in os.listdir(dir_path):
    full_path = os.path.join(dir_path, file)
    if os.path.isfile(full_path):
        df = pd.read_csv(full_path)
        attendance = len(df)
        if attendance >= max_attendance:
            max_attendance = attendance
            max_date = file

print(max_date)
```
