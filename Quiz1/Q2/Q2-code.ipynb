```python
%pip install pypdf
from pypdf import PdfReader
import matplotlib.pyplot as plt
import pandas as pd

# My resume file
my_resume = "./Lea Frost Resume Spring 2025.pdf"
```

```python
# load in stop words
stop_file_name = "NLTK's20of20stopwords.txt"
# Read stop file
stop_file = open(stop_file_name)
stop_text = stop_file.read()
stop_file.close
stop_words = stop_text.split()
```

Read resume functions

```python
# Read resume, return as list of words
def read_resume(resume_name):
    reader = PdfReader(resume_name)
    pages = reader.pages[0]
    text = pages.extract_text().lower()

    # Split into words
    resume_words = text.split()

    # Clean data by removing non-alphabetical values
    resume_words = [w for w in resume_words if w.isalpha()]

    return resume_words

# Remove stop words, return specific word list
def remove_stop(resume_words, stop_words):
    specific_words = [w for w in resume_words if not w in stop_words]
    return specific_words

def get_top_20(words):
    # Top 20 words
    top_20 = pd.Series(words).value_counts().head(20)
    return top_20
```

Plot top 20 words (before removing stop words)

```python
resume_words = read_resume(my_resume)
top_20 = get_top_20(resume_words)
# Plot histogram
top_20.plot(kind="bar", width=0.8)
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.title('Resume Word Histogram')
plt.xticks(rotation=45, ha='right')
plt.show()
```

Plot 20 specific words

```python
# Top 20 specific words
specific_words = remove_stop(resume_words, stop_words)
top_20_specific = get_top_20(specific_words)
# Plot histogram
top_20_specific.plot(kind="bar", width=0.8)
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.title('Resume Specific Word Histogram')
plt.xticks(rotation=45, ha='right')
plt.show()
```

#### Analyze: Note which words emerge now. Was removing stop_words helpful in revealing more about you (from the resume)

Removing stop words was very helpful in revealing more information about me. The previous list included only vague discriptors like computer, learning, university, etc. Removing stop words revealed more specific information such as networks, statistics. programming. etc

Task 2: Read words from all resumes

```python
import os

# Collect files
data_folder = "./data"
files = [os.path.join(data_folder,f) for f in os.listdir(data_folder) if os.path.isfile(os.path.join(data_folder,f))]
print(files)
```

```python
# get all resume words
resume_words = []
for file in files:
    resume_words += read_resume(file)

# top 20 words
top_20_all = get_top_20(resume_words)
```

```python
from matplotlib.pyplot import figure

# plot histogram 
top_20_all.plot(kind="bar", width=0.8)
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.title('Class Resumes Non-Specific Word Histogram')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
specific_words = remove_stop(resume_words, stop_words)
top_20_specific_all = get_top_20(specific_words)

# plot histogram 
top_20_specific_all.plot(kind="bar", width=0.8)
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.title('Class Resumes Specific Word Histogram')
plt.xticks(rotation=45, ha='right')
plt.show()
```

##### Analyze: Note which words emerge now. Was removing stop_words helpful in revealing more about the class (from the resumes).

Removing stop words allowed more informative words to populate the top 20. Compared to the non-specific list, we can see a lot more skills that students list on their resume such as engineering, programming, design, etc. These did not appear in the previous list.
